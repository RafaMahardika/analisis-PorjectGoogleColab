```python
import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

tugas Analisis 1:

Jika kita mencoba [5000, 7000, 3000] * 0.9 pada list biasa, Python akan menghasilkan TypeError / Error karena list Python tidak mendukung operasi matematika langsung (vectorized operations) dengan angka desimal (float).

Mengapa berbeda? Array NumPy dirancang khusus untuk komputasi matematika, sehingga perkalian langsung diterapkan ke setiap elemen di dalamnya (element-wise). Sedangkan pada list Python biasa, operator * hanya bisa digunakan dengan angka bulat (int) untuk mengulang isi list (duplikasi).

```python
import pandas as pd

data_kantin = {
    'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}

df_latihan = pd.DataFrame(data_kantin)
print(df_latihan)
```

Tugas Analisis 2:

Kolom yang memiliki data kosong (None/NaN) adalah kolom menu (baris index ke-4) dan kolom terjual (baris index ke-2).

Risikonya: Jika data langsung dianalisis tanpa dibersihkan, perhitungan statistik (seperti rata-rata atau total penjualan) bisa menjadi tidak akurat atau menghasilkan angka NaN. Selain itu, saat membuat visualisasi data/grafik, data kosong bisa menyebabkan error atau informasi yang salah.

```python
df = pd.read_csv('data_kantin.csv')

print("--- 5 Baris Pertama ---")
print(df.head())

print("\n--- Info Data ---")
df.info()

print("\n--- Ringkasan Statistik ---")
print(df.describe())

print("\n--- Ukuran Data ---")
print(df.shape)
```

Tugas Analisis 3:

Kolom yang jumlah non-null-nya lebih sedikit dari total baris (110 baris) adalah kolom menu (hanya 107 non-null) dan kolom terjual (hanya 104 non-null).

Artinya: Terdapat missing values (data kosong/hilang) pada kedua kolom tersebut. Kolom menu memiliki 3 data kosong, dan kolom terjual memiliki 6 data kosong.

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Tugas Analisis 4:

Kolom terjual diisi dengan fillna(0) karena merupakan data numerik (jumlah barang). Asumsinya, jika tidak dicatat/kosong, berarti tidak ada barang yang terjual pada hari itu (0 unit), sehingga data transaksi lain pada baris tersebut masih bisa dianalisis.

Kolom menu dihapus dengan dropna() karena kolom menu adalah identitas utama dari transaksi (kategori/nama produk). Jika nama produknya kosong, data transaksi tersebut menjadi tidak valid dan tidak bisa diketahui barang apa yang dijual.

```python
print("Jumlah baris duplikat:", df.duplicated().sum())
df = df.drop_duplicates()

df['harga'] = df['harga'].astype(int)
print(df.dtypes)
```

```python
import pandas as pd
df = pd.read_csv('data_kantin.csv')
df.head()
df.info()
df.describe()
```

```python
# 1. Hapus duplikat
df = df.drop_duplicates()

# 2. Isi missing value (harus dieksekusi agar non-null count jadi sama)
df['menu'] = df['menu'].fillna('Tidak Diketahui')
df['terjual'] = df['terjual'].fillna(0)
# 3. Ubah tipe data
df['tanggal'] = pd.to_datetime(df['tanggal'])
df['terjual'] = df['terjual'].astype(int)

# 4. Cek hasil akhir
df.info()
```

```python
# 1. Kolom Turunan: Menghitung total omset/pendapatan per transaksi
df['total_pendapatan'] = df['harga'] * df['terjual']

# 2. Filtering: Mengambil transaksi dengan pendapatan di atas 50.000
df_filtered = df[df['total_pendapatan'] > 50000]

# 3. Sorting: Mengurutkan dari transaksi dengan pendapatan tertinggi
df_sorted = df.sort_values(by='total_pendapatan', ascending=False)

# 4. Groupby / Agregasi: Total pendapatan & total unit terjual berdasarkan Kategori
df_grouped = df.groupby('kategori')[['terjual', 'total_pendapatan']].sum().reset_index()

# Tampilkan hasil pengelompokan
df_grouped
```
